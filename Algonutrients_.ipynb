```python
# NUTRILINK AI
# STEP 9A: PREPARE MACHINE LEARNING DATA

from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_val_predict
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Select predictor features
predictor_features = [
    "FATCE(g)",
    "FIBTG(g)",
    "CHOAVLDF(g)",
    "ENERC(kcal)",
    "CA(mg)",
    "FE(mg)",
    "ZN(mg)",
    "MG(mg)",
    "LYS(mg)",
    "MET(mg)",
    "LEU(mg)"
]

# Create input variables (X)
X = ingredient_df[predictor_features]

# Create target variable (y)
y = ingredient_df["PROTCNT(g)"]

# Display dataset dimensions
print("Number of ingredient records:", X.shape[0])
print("Number of predictor features:", X.shape[1])

print("\nTarget variable:")
print("Protein content (PROTCNT(g))")

print("\nData preparation completed!")
```

```python
# NUTRILINK AI
# STEP 9A: PREPARE MACHINE LEARNING DATA

import pandas as pd
import numpy as np

from sklearn.model_selection import KFold, cross_val_predict
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.svm import SVR

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Select predictor features
predictor_features = [
    "FATCE(g)",
    "FIBTG(g)",
    "CHOAVLDF(g)",
    "ENERC(kcal)",
    "CA(mg)",
    "FE(mg)",
    "ZN(mg)",
    "MG(mg)",
    "LYS(mg)",
    "MET(mg)",
    "LEU(mg)"
]

# Input features
X = ingredient_df[predictor_features]

# Target variable
y = ingredient_df["PROTCNT(g)"]

print("Records:", X.shape[0])
print("Predictor features:", X.shape[1])
print("Data preparation completed!")
```

```python
# NUTRILINK AI
# STEP 9B: COMPARE FIVE MACHINE LEARNING MODELS

models = {
    "Linear Regression": make_pipeline(
        StandardScaler(),
        LinearRegression()
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=42
    ),

    "Support Vector Regression": make_pipeline(
        StandardScaler(),
        SVR(kernel="rbf", C=10, epsilon=0.1)
    )
}

# Use the same cross-validation splits for all models
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

results = []

for name, model in models.items():

    # Generate cross-validation predictions
    predictions = cross_val_predict(
        model,
        X,
        y,
        cv=cv
    )

    # Calculate evaluation metrics
    mae = mean_absolute_error(y, predictions)
    rmse = np.sqrt(mean_squared_error(y, predictions))
    r2 = r2_score(y, predictions)

    results.append({
        "Model": name,
        "MAE": mae,
        "RMSE": rmse,
        "R2 Score": r2
    })

# Create comparison table
comparison_df = pd.DataFrame(results)

comparison_df = comparison_df.sort_values(
    by="MAE",
    ascending=True
).reset_index(drop=True)

print("NUTRILINK AI - MODEL PERFORMANCE COMPARISON")
print("--------------------------------------------")

display(comparison_df.round(4))
```

```python
# NUTRILINK AI
# STEP 9C: COMPARE MODEL PERFORMANCE VISUALLY

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(
    comparison_df["Model"],
    comparison_df["MAE"]
)

plt.title("Comparison of Machine Learning Models")
plt.xlabel("Machine Learning Model")
plt.ylabel("Mean Absolute Error (g)")

plt.xticks(rotation=25, ha="right")
plt.tight_layout()
plt.show()
```

```python
# NUTRILINK AI
# STEP 9E: MODEL COMPARISON WITHOUT DIRECT PROTEIN-RELATED FEATURES

from sklearn.model_selection import KFold, cross_val_predict
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.svm import SVR
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

import numpy as np
import pandas as pd

# Use a more conservative set of predictor features
independent_features = [
    "FATCE(g)",
    "FIBTG(g)",
    "CA(mg)",
    "FE(mg)",
    "ZN(mg)",
    "MG(mg)"
]

# Prepare input and target
X_clean = ingredient_df[independent_features]
y_clean = ingredient_df["PROTCNT(g)"]

# Define the five models
models_clean = {
    "Linear Regression": make_pipeline(
        StandardScaler(),
        LinearRegression()
    ),

    "Decision Tree": DecisionTreeRegressor(
        max_depth=3,
        random_state=42
    ),

    "Random Forest": RandomForestRegressor(
        n_estimators=200,
        random_state=42
    ),

    "Gradient Boosting": GradientBoostingRegressor(
        n_estimators=100,
        learning_rate=0.05,
        max_depth=2,
        random_state=42
    ),

    "Support Vector Regression": make_pipeline(
        StandardScaler(),
        SVR(kernel="rbf", C=10, epsilon=0.1)
    )
}

# Use identical cross-validation splits
cv = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Compare models
clean_results = []

for name, model in models_clean.items():

    predictions = cross_val_predict(
        model,
        X_clean,
        y_clean,
        cv=cv
    )

    clean_results.append({
        "Model": name,
        "MAE": mean_absolute_error(y_clean, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_clean, predictions)),
        "R2 Score": r2_score(y_clean, predictions)
    })

# Display results
clean_comparison_df = pd.DataFrame(clean_results)

clean_comparison_df = clean_comparison_df.sort_values(
    by="MAE"
).reset_index(drop=True)

print("NUTRILINK AI: SECOND MODEL COMPARISON")
print("=" * 55)

for _, row in clean_comparison_df.iterrows():
    print(f"\nModel: {row['Model']}")
    print(f"MAE: {row['MAE']:.4f}")
    print(f"RMSE: {row['RMSE']:.4f}")
    print(f"R2 Score: {row['R2 Score']:.4f}")
```

```python
# NUTRILINK AI
# STEP 18: FIRST PROTEIN FORMULATION

# Ingredient quantities in grams
formulation = {
    "Soya bean": 40,
    "Cowpea": 30,
    "Pigeon pea": 20,
    "Bambara groundnut": 10
}

results = []

for ingredient, weight in formulation.items():

    # Get the ingredient's nutritional information
    row = nutrilink_shortlist[
        nutrilink_shortlist["Ingredient"] == ingredient
    ].iloc[0]

    protein_per_100g = row["Protein_g"]
    fat_per_100g = row["Fat_g"]
    fibre_per_100g = row["Fibre_g"]
    energy_per_100g = row["Energy_kcal"]

    # Calculate nutrient contribution
    protein = weight * protein_per_100g / 100
    fat = weight * fat_per_100g / 100
    fibre = weight * fibre_per_100g / 100
    energy = weight * energy_per_100g / 100

    results.append({
        "Ingredient": ingredient,
        "Weight_g": weight,
        "Protein_g": protein,
        "Fat_g": fat,
        "Fibre_g": fibre,
        "Energy_kcal": energy
    })

formulation_result = pd.DataFrame(results)

print("NUTRILINK AI - FIRST FORMULATION")
print("=" * 65)

print(formulation_result.round(2).to_string(index=False))

print("\nTOTAL NUTRITIONAL COMPOSITION")
print("-" * 65)

print("Total mixture weight:",
      formulation_result["Weight_g"].sum(), "g")

print("Total protein:",
      round(formulation_result["Protein_g"].sum(), 2), "g")

print("Total fat:",
      round(formulation_result["Fat_g"].sum(), 2), "g")

print("Total fibre:",
      round(formulation_result["Fibre_g"].sum(), 2), "g")

print("Total energy:",
      round(formulation_result["Energy_kcal"].sum(), 2), "kcal")
```

```python
# NUTRILINK AI
# STEP 19A: IMPORT OPTIMIZATION LIBRARIES

import numpy as np
import pandas as pd

from scipy.optimize import minimize, linprog

print("Optimization libraries imported successfully!")
```

```python
# NUTRILINK AI
# STEP 19B: FORMULATION REQUIREMENTS

TARGET_PROTEIN = 25       # Minimum protein required in grams
BATCH_SIZE = 100          # Total mixture weight in grams
MIN_INGREDIENT = 0        # Minimum quantity of each ingredient
MAX_INGREDIENT = 40       # Maximum quantity of each ingredient

# Nutritional data
formulation_data = nutrilink_shortlist.copy()

ingredients = formulation_data["Ingredient"].tolist()

protein_values = formulation_data["Protein_g"].to_numpy()

number_of_ingredients = len(ingredients)

# Equal distribution as the starting reference
equal_weight = BATCH_SIZE / number_of_ingredients

print("FORMULATION REQUIREMENTS")
print("=" * 45)
print("Target protein:", TARGET_PROTEIN, "g")
print("Total mixture weight:", BATCH_SIZE, "g")
print("Number of ingredients:", number_of_ingredients)
print("Maximum ingredient inclusion:", MAX_INGREDIENT, "g")
print("Equal starting proportion:", round(equal_weight, 2), "g")
```

```python
# NUTRILINK AI
# STEP 19C: CHECK FORMULATION FEASIBILITY

maximum_protein_solution = linprog(
    c=-protein_values,
    A_eq=[np.ones(number_of_ingredients)],
    b_eq=[BATCH_SIZE],
    bounds=[
        (MIN_INGREDIENT, MAX_INGREDIENT)
        for _ in ingredients
    ],
    method="highs"
)

if maximum_protein_solution.success:

    maximum_possible_protein = (
        -maximum_protein_solution.fun / 100
    )

    print("MAXIMUM POSSIBLE PROTEIN")
    print("=" * 45)
    print(
        "Maximum protein in mixture:",
        round(maximum_possible_protein, 2),
        "g"
    )

    if TARGET_PROTEIN > maximum_possible_protein:
        print("The selected protein target is not achievable.")
    else:
        print("The selected protein target is achievable.")

else:
    print("Unable to calculate formulation feasibility.")
    print(maximum_protein_solution.message)
```

```python
# NUTRILINK AI
# STEP 19D: AUTOMATIC FORMULATION OPTIMIZATION

def formulation_objective(weights):

    # Minimize deviation from equal ingredient proportions
    return np.sum((weights - equal_weight) ** 2)


def protein_constraint(weights):

    # Calculate total protein in the mixture
    total_protein = np.sum(
        weights * protein_values / 100
    )

    return total_protein - TARGET_PROTEIN


initial_weights = np.full(
    number_of_ingredients,
    equal_weight
)

constraints = [
    {
        "type": "eq",
        "fun": lambda weights: np.sum(weights) - BATCH_SIZE
    },
    {
        "type": "ineq",
        "fun": protein_constraint
    }
]

bounds = [
    (MIN_INGREDIENT, MAX_INGREDIENT)
    for _ in ingredients
]

if maximum_protein_solution.success and TARGET_PROTEIN <= maximum_protein_solution.fun * -1 / 100:

    optimization_result = minimize(
        formulation_objective,
        initial_weights,
        method="SLSQP",
        bounds=bounds,
        constraints=constraints,
        options={
            "maxiter": 1000,
            "ftol": 1e-9
        }
    )

    if optimization_result.success:

        optimized_weights = optimization_result.x

        print("FORMULATION OPTIMIZATION SUCCESSFUL!")

    else:
        print("Optimization failed:")
        print(optimization_result.message)

else:
    print("Cannot optimize because the protein target is infeasible.")
```

```python
# NUTRILINK AI
# STEP 19E: DISPLAY OPTIMIZED INGREDIENT COMBINATION

if optimization_result.success:

    optimized_formulation = pd.DataFrame({
        "Ingredient": ingredients,
        "Weight_g": optimized_weights,
        "Protein_per_100g": protein_values
    })

    optimized_formulation["Protein_Contribution_g"] = (
        optimized_formulation["Weight_g"]
        * optimized_formulation["Protein_per_100g"]
        / 100
    )

    print("NUTRILINK AI - OPTIMIZED FORMULATION")
    print("=" * 75)

    print(
        optimized_formulation.round(3).to_string(index=False)
    )

    print("\nTOTAL MIXTURE WEIGHT:",
          round(optimized_weights.sum(), 2), "g")

    print("TOTAL PROTEIN:",
          round(
              optimized_formulation[
                  "Protein_Contribution_g"
              ].sum(), 2
          ), "g")
```

```python
# NUTRILINK AI
# STEP 20A: COMPLETE NUTRITIONAL ANALYSIS

nutrient_columns = [
    "Protein_g",
    "Fat_g",
    "Fibre_g",
    "Energy_kcal"
]

# Retrieve nutritional data for each selected ingredient
nutrition_lookup = nutrilink_shortlist.set_index("Ingredient")

nutritional_results = optimized_formulation[
    ["Ingredient", "Weight_g"]
].copy()

for nutrient in nutrient_columns:

    nutritional_results[nutrient] = (
        nutritional_results["Ingredient"].map(
            nutrition_lookup[nutrient]
        ) * nutritional_results["Weight_g"] / 100
    )

print("NUTRILINK AI - NUTRITIONAL CONTRIBUTIONS")
print("=" * 75)

print(nutritional_results.round(3).to_string(index=False))
```

```python
# NUTRILINK AI
# STEP 22A: USER-DEFINED FORMULATION OPTIMIZER

from scipy.optimize import minimize, linprog
import numpy as np
import pandas as pd

def optimize_nutrilink(
    target_protein=25,
    batch_size=100,
    max_inclusion=40
):

    data = nutrilink_shortlist.copy()

    ingredients = data["Ingredient"].tolist()

    protein = data["Protein_g"].to_numpy(dtype=float)

    fat = data["Fat_g"].to_numpy(dtype=float)

    fibre = data["Fibre_g"].to_numpy(dtype=float)

    energy = data["Energy_kcal"].to_numpy(dtype=float)

    n = len(ingredients)

    # Check whether the target is achievable
    feasibility = linprog(
        c=-protein,
        A_eq=[np.ones(n)],
        b_eq=[batch_size],
        bounds=[(0, max_inclusion)] * n,
        method="highs"
    )

    if not feasibility.success:
        raise ValueError(
            "Unable to find a feasible formulation with these constraints."
        )

    maximum_protein = -feasibility.fun / 100

    if target_protein > maximum_protein:
        raise ValueError(
            f"Target is too high. Maximum achievable protein is "
            f"{maximum_protein:.2f} g."
        )

    # Objective: keep proportions as close as possible to equal shares
    equal_weight = batch_size / n

    def objective(weights):
        return np.sum((weights - equal_weight) ** 2)

    def protein_constraint(weights):
        return np.sum(weights * protein / 100) - target_protein

    constraints = [
        {
            "type": "eq",
            "fun": lambda weights: np.sum(weights) - batch_size
        },
        {
            "type": "ineq",
            "fun": protein_constraint
        }
    ]

    initial_weights = np.full(n, equal_weight)

    result = minimize(
        objective,
        initial_weights,
        method="SLSQP",
        bounds=[(0, max_inclusion)] * n,
        constraints=constraints,
        options={"maxiter": 2000, "ftol": 1e-10}
    )

    if not result.success:
        raise RuntimeError(
            f"Optimization failed: {result.message}"
        )

    weights = result.x

    # Construct the formulation table
    output = pd.DataFrame({
        "Ingredient": ingredients,
        "Weight_g": weights,
        "Protein_g": weights * protein / 100,
        "Fat_g": weights * fat / 100,
        "Fibre_g": weights * fibre / 100,
        "Energy_kcal": weights * energy / 100
    })

    # Remove negligible quantities
    output = output[output["Weight_g"] > 0.001].copy()

    totals = {
        "Batch_Weight_g": output["Weight_g"].sum(),
        "Protein_g": output["Protein_g"].sum(),
        "Fat_g": output["Fat_g"].sum(),
        "Fibre_g": output["Fibre_g"].sum(),
        "Energy_kcal": output["Energy_kcal"].sum(),
        "Maximum_Possible_Protein_g": maximum_protein
    }

    return output.reset_index(drop=True), totals
```

```python
# NUTRILINK AI
# STEP 22B: TEST MULTIPLE PROTEIN TARGETS

protein_targets = [20, 23, 25]

formulation_scenarios = {}

for target in protein_targets:

    formulation, totals = optimize_nutrilink(
        target_protein=target,
        batch_size=100,
        max_inclusion=40
    )

    formulation_scenarios[target] = {
        "formulation": formulation,
        "totals": totals
    }

    print("\n" + "=" * 75)
    print(f"FORMULATION FOR {target}g PROTEIN")
    print("=" * 75)

    print(
        formulation.round(2).to_string(index=False)
    )

    print("\nTotal weight:",
          round(totals["Batch_Weight_g"], 2), "g")

    print("Total protein:",
          round(totals["Protein_g"], 2), "g")

    print("Total fat:",
          round(totals["Fat_g"], 2), "g")

    print("Total fibre:",
          round(totals["Fibre_g"], 2), "g")

    print("Total energy:",
          round(totals["Energy_kcal"], 2), "kcal")
```
